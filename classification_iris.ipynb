```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score
import pandas as pd
```

```python
# dataset 
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
data = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
data.head(5)
```

```python
data['target'] = load_iris().target
```

```python
data.head()
```

```python
data.tail()
```

```python
print(data['sepal length (cm)'].min(), data['sepal length (cm)'].max())
print(data['sepal width (cm)'].min(), data['sepal width (cm)'].max())
print(data['petal length (cm)'].min(), data['petal length (cm)'].max())
print(data['petal width (cm)'].min(), data['petal width (cm)'].max())
```

```python
# splitting the dataset 
x = data.drop('target', axis=1)
y = data['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

**LDA**

```python
lda = LinearDiscriminantAnalysis()
lda.fit(x_train, y_train)
```

```python
y_test_pred = lda.predict(x_test)
y_train_pred = lda.predict(x_train)
```

```python
y_test_pred
```

```python
# evaluation of the model 
print("Accuracy score:- ",accuracy_score(y_test, y_test_pred))
print("confusion_matrix:- ",confusion_matrix(y_test, y_test_pred))
print("F1 score:- ",f1_score(y_test, y_test_pred, average='weighted'))
print("precision_score:- ",precision_score(y_test, y_test_pred, average='weighted'))
print("Recall score:- ",recall_score(y_test, y_test_pred, average='weighted'))
# perfect score is 1.0 which means the model is performing great
```

**Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression 
import joblib # saving the model 

lg = LogisticRegression(solver='lbfgs')
lg.fit(x_train, y_train)
y_test_pred = lg.predict(x_test)
y_train_pred = lg.predict(x_train)
joblib.dump(lg, 'iris_model.joblib') # saving the model in the current directory 

print("accuracy_score:- ", accuracy_score(y_test, y_test_pred))
print("confusion_matrix:- ",confusion_matrix(y_test, y_test_pred))
print("F1 score:- ",f1_score(y_test, y_test_pred, average='weighted'))
print("precision_score:- ", precision_score(y_test, y_test_pred, average='weighted'))
print("recall_score:- ",recall_score(y_test, y_test_pred, average='weighted'))
# perfect score is 1.0 which means the model is performing great
```
